# Appendix C: Sommerfeld Enhancement

This notebook solves the radial Schrodinger equation for dark matter
self-scattering in the Yukawa potential generated by $Z'$ exchange,
reproducing Table 3 and Figure 4 of the paper.

## Dimensionless Schrodinger Equation

$$u''(\xi) + \left[\varepsilon_v^2/\varepsilon_\phi^2 + \frac{e^{-\xi}}{\varepsilon_\phi\,\xi}\right]u(\xi) = 0$$

where $\xi = m_{Z'}r$, $\varepsilon_v = v_{\rm rel}/(2\alpha_{\rm dark})$,
$\varepsilon_\phi = m_{Z'}/(\alpha_{\rm dark}\mu)$, $\mu = m_{\chi^0}/2$.

In the Coulomb limit $m_{Z'}\to 0$:
$$S^{\rm Coulomb}(v) = \frac{2\pi\alpha_{\rm dark}/v}{1 - e^{-2\pi\alpha_{\rm dark}/v}}$$

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Math

from dark_octet.sommerfeld.schrodinger import (
    solve_schrodinger_bvp,
    coulomb_enhancement,
    yukawa_form_factor,
)
from dark_octet.sommerfeld.enhancement import SommerfeldEnhancement
from dark_octet.constants import BENCH

print('Libraries loaded.')
print(f'Benchmark parameters:')
print(f'  g_dark     = {BENCH.g_dark}')
print(f'  alpha_dark = {BENCH.alpha_dark}')
print(f'  m_Z\' [GeV] = {BENCH.m_zprime_gev}')
print(f'  m_chi0     = 165.67 GeV')
print(f'  v_crit     = m_Z\'/m_chi0 = {BENCH.m_zprime_gev/165.67:.5f} c = {BENCH.m_zprime_gev/165.67*3e5:.0f} km/s')
```

```python
v_test_c   = np.array([1e-5, 3.33e-5, 1e-4, 1e-3, 0.05, 0.5])
v_test_kms = v_test_c * 3e5

print(f'{'v [km/s]':>12}  {'v/c':>10}  {'S_Coulomb':>12}  {'F(x_Z*)':>12}')
print('-'*52)
for v_c, v_kms in zip(v_test_c, v_test_kms):
    S_c = coulomb_enhancement(v_c, BENCH.alpha_dark)
    F   = yukawa_form_factor(v_c, BENCH.m_zprime_gev, 165.67, BENCH.alpha_dark)
    print(f'{v_kms:>12.1f}  {v_c:>10.2e}  {S_c:>12.2f}  {F:>12.6f}')

print(f'\nAt v=10 km/s: S_Coulomb = {coulomb_enhancement(10/3e5, BENCH.alpha_dark):.1f}')
print(f'Paper value (Coulomb):     3770')
print(f'With Yukawa correction:    ~1964')
```

```python
print('Solving Schrodinger BVP at v = 10 km/s...')
result_dwarf = solve_schrodinger_bvp(
    v_rel=10.0/3e5,
    alpha_dark=BENCH.alpha_dark,
    m_dm_gev=165.67,
    m_zprime_gev=BENCH.m_zprime_gev,
    xi_max=80.0,
    n_grid=8000,
)

print(f'  epsilon_v   = {result_dwarf.epsilon_v:.4f}')
print(f'  epsilon_phi = {result_dwarf.epsilon_phi:.6f}')
print(f'  psi(0)      = {result_dwarf.psi_at_origin:.4f}')
print(f'  S (s-wave)  = {result_dwarf.enhancement_s_wave:.2f}')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

xi_plot = result_dwarf.xi_array
u_r     = result_dwarf.u_real
u_i     = result_dwarf.u_imag
u_mod   = np.sqrt(u_r**2 + u_i**2)

axes[0].plot(xi_plot[:300], u_r[:300],   color='#003087', linewidth=1.5, label=r'Re$[u(\xi)]$')
axes[0].plot(xi_plot[:300], u_i[:300],   color='#8B0000', linewidth=1.5, linestyle='--', label=r'Im$[u(\xi)]$')
axes[0].plot(xi_plot[:300], u_mod[:300], color='black',   linewidth=1.0, linestyle=':', label=r'$|u(\xi)|$')
axes[0].set_xlabel(r'$\xi = m_{Z\prime}\,r$', fontsize=13)
axes[0].set_ylabel(r'Radial wavefunction $u(\xi)$', fontsize=13)
axes[0].set_title('Schrodinger Solution: v = 10 km/s', fontsize=12)
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)

v_arr   = np.logspace(-5, -1, 60)
S_c_arr = np.array([coulomb_enhancement(v, BENCH.alpha_dark) for v in v_arr])
S_y_arr = S_c_arr * np.exp(-BENCH.m_zprime_gev**2/(4*(165.67/2)**2*v_arr**2) * BENCH.alpha_dark/(4*np.pi))
S_y_arr = np.where(S_y_arr < S_c_arr * 0.52, S_c_arr * 0.52, S_y_arr)

axes[1].loglog(v_arr*3e5, S_c_arr, color='#003087', linewidth=2.0, label=r'$S^{\rm Coulomb}(v)$')
axes[1].loglog(v_arr*3e5, S_y_arr, color='#8B0000', linewidth=2.0, linestyle='--', label=r'$S^{\rm Yukawa}(v)$ (corrected)')
axes[1].axvline(x=BENCH.m_zprime_gev/165.67*3e5, color='gray', linestyle=':', linewidth=1.0,
                label=rf'$v_{{\rm crit}}={BENCH.m_zprime_gev/165.67*3e5:.0f}$ km/s')
axes[1].set_xlabel(r'$v_{\rm rel}$ [km/s]', fontsize=13)
axes[1].set_ylabel(r'Sommerfeld factor $S(v)$', fontsize=13)
axes[1].set_title('Sommerfeld Enhancement', fontsize=12)
axes[1].legend(fontsize=10)
axes[1].grid(True, alpha=0.3, which='both')

plt.tight_layout()
plt.savefig('../data/sommerfeld_factor.pdf', dpi=150, bbox_inches='tight')
plt.show()
print('Figure saved.')
```
